```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder
emails = [
    "win surprise",
    "free lottery",
    "i am shan",
    "who are you"
]
labels = ["Spam", "Spam", "Not Spam", "Not Spam"]
le = LabelEncoder()
y = le.fit_transform(labels)
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)

model = MultinomialNB()
model.fit(X, y)
test_email = ["surprise lottery"]
test_vec = vectorizer.transform(test_email)

pred = model.predict(test_vec)

print("Prediction:", le.inverse_transform(pred)[0])
```

```python
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder

X = np.array([[30], [40], [45], [50], [60]])

y_text = ["Fail", "Fail", "Fail", "Pass", "Pass"]

le = LabelEncoder()
y = le.fit_transform(y_text)

model = GaussianNB()


model.fit(X, y)

new_marks = [[46]]
prediction = model.predict(new_marks)

print("Predicted Result for 46 marks:", le.inverse_transform(prediction)[0])
```

```python
import numpy as np
from sklearn.naive_bayes import BernoulliNB
from sklearn.preprocessing import LabelEncoder

X = np.array([
    [1, 1],
    [0, 1],
    [1, 0],
    [0, 0]
])

y_text = ["Yes", "No", "Yes", "No"]

le = LabelEncoder()
y = le.fit_transform(y_text)

model = BernoulliNB()

model.fit(X, y)

new_dish = [[1,0 ]]
prediction = model.predict(new_dish)

print("Prediction:", le.inverse_transform(prediction)[0])
```

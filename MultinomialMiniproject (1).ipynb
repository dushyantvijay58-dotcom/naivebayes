```python
x=df[['comment','brand']]
y=df['tonality']


x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
l=LabelEncoder()
y=l.fit_transform(y)
```

```python
vectrizor=CountVectorizer()
x=vectrizor.fit_transform(x_train)
```

```python
x_train_vec = vectrizor.fit_transform(x_train['comment'])


model = MultinomialNB()
model.fit(x_train_vec, y_train)
```

```python
acc=model.score(x_train_vec,y_train)
acc
```

```python
a=vectrizor.transform(x_test['comment'])
y_pred=model.predict(a)
```
